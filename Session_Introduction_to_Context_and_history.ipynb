### Langchain as Orchestrator

```python
# Note: You may need to run `pip install langchain-google-genai langchain` first
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableLambda

# 1. Initialize the LLM wrapper

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", google_api_key=google_key)


# llm = ChatOpenAI(model="gpt-4o-mini", api_key=openai_key)

# Define the prompt
prompt_template = PromptTemplate.from_template("Imagine you're  a travel expert, answer the user query in one line - {user_message}")


# 3. Wrap our sanitize function so LangChain can use it in a pipeline
def sanitize_lc_output(ai_message) -> str:
    text = ai_message.content if hasattr(ai_message, 'content') else str(ai_message)
    return text.strip().replace("**", "")

chain = prompt_template|llm|RunnableLambda(sanitize_lc_output)
```

## Replaing Lists with MessageHistory

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser
from langchain_core.chat_history import InMemoryChatMessageHistory

## Making the History
# 1. Initialize our components
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", google_api_key=api_key)
# Define your parser
parser = StrOutputParser()

# 2. Define the template with MessagesPlaceholder
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant. Keep answers to one line."),
    # Inject history here with a list called as chat_history
    MessagesPlaceholder(variable_name = 'chat_history'),
    ("human", "{user_input}")
])
# Replcing list with chatmessagehistory
past_conversation_list = InMemoryChatMessageHistory()
# 3. Build the Chain
chain = prompt_template | llm | parser
while True:
  user_text = input("Enter your message: ")
  if(user_text.lower() == 'exit'):
    break
  ai_msg = chain.invoke({'chat_history':past_conversation_list.messages, 'user_input': user_text })
  print(ai_msg)
  past_conversation_list.add_user_message(user_text)
  past_conversation_list.add_ai_message(ai_msg)
```

```python
# We started with AIMEssage, HumanMessage, we were appending AIMessage and HumanMessage to list and then making prompt
# Problem: We had to update chain again and again with the new prompt

# We swtiched to MessagePlaceholder to hold our history, now we don't need to make a prompt we just use a prompttemplate, we plug in the history with invoke?
# Problem: Lists

# We switched to InMemoryChatMessageHistory instead of lists
# Problem : we're still doing the append part ourselves, langchain is the orchestrator, it should do it.
```

# The Ultimate Chatbot: Automating Memory

We have a working loop, but we are still manually updating the memory at the bottom of it. If we forget those lines of code, our chatbot breaks.

To fix this, LangChain provides a magical wrapper called **`RunnableWithMessageHistory`**.

[Image of LangChain RunnableWithMessageHistory wrapper automatically intercepting inputs and outputs to update the chat database]

This wrapper sits on the outside of our chain. When we send a message:
1. It automatically looks up the past history based on a `session_id`.
2. It injects that history into our `MessagesPlaceholder`.
3. It runs the LLM.
4. It catches the LLM's output and **automatically saves both the user query and the AI response to our history object.**

Let's look at how clean our loop becomes when we let LangChain do all the work.

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, AIMessage
from langchain_core.output_parsers import StrOutputParser
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

session_store = {}
def get_session_history(session_id: str):
  if(session_id not in session_store):
    session_store[session_id] = InMemoryChatMessageHistory()
  return session_store[session_id]

## Making the History
# 1. Initialize our components
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", google_api_key=api_key)
# Define your parser
parser = StrOutputParser()
```

```python

# 2. Define the template with MessagesPlaceholder
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant. Keep answers to one line."),
    # Inject history here with a list called as chat_history
    MessagesPlaceholder(variable_name = 'chat_history'),
    ("human", "{user_input}")
])
# Replcing list with chatmessagehistory
# past_conversation_list = InMemoryChatMessageHistory()
# 3. Build the Chain
chain = prompt_template | llm | parser

#Upgrade the chain
chain_with_memory = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key= "user_input",
    history_messages_key="chat_history"
)
user_id = input("Please Enter your user id: ")
config = {"configurable": {"session_id": user_id}}
while True:
  user_text = input("Enter your message: ")
  if(user_text.lower() == 'exit'):
    break
  ai_msg = chain_with_memory.invoke({'user_input': user_text }, config = config)
  print(ai_msg)
  # past_conversation_list.add_user_message(user_text)
  # past_conversation_list.add_ai_message(ai_msg)
```
